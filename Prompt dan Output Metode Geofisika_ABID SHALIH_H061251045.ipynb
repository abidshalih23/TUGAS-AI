## 1. Struktur Prompt

### Role
Anda adalah seorang ahli geofisika yang berpengalaman dalam metode seismik refraksi dan interpretasi data bawah permukaan.

### Task
Jelaskan dan analisis penerapan metode seismik refraksi untuk mengetahui kondisi bawah permukaan berdasarkan data waktu tiba gelombang seismik pada beberapa geofon. Tentukan kecepatan gelombang pada setiap lapisan, estimasikan kedalaman antarmuka lapisan, dan jelaskan interpretasi geologi/geofisikanya.

### Context
Metode seismik refraksi memanfaatkan gelombang seismik yang dibiaskan secara kritis pada batas antara dua lapisan yang memiliki perbedaan kecepatan. Metode ini dapat digunakan untuk memperkirakan struktur lapisan bawah permukaan, misalnya ketebalan lapisan pelapukan dan posisi batuan dasar. Analisis harus menggunakan konsep waktu tempuh, kecepatan gelombang, gelombang langsung, dan gelombang refraksi.

### Input
Gunakan data berikut sebagai contoh:

- Jumlah lapisan: 2 lapisan bawah permukaan.
- Kecepatan lapisan pertama: sekitar 500 m/s.
- Kecepatan lapisan kedua: sekitar 1500 m/s.
- Jarak antar geofon: 5 m.
- Data waktu tiba gelombang tersedia untuk setiap posisi geofon.
- Jika data numerik belum tersedia, jelaskan metode perhitungannya secara simbolik dan berikan contoh perhitungan sederhana.

### Constraints
1. Gunakan bahasa Indonesia yang mudah dipahami oleh mahasiswa S1 Geofisika.
2. Jangan hanya memberikan hasil akhir; jelaskan konsep dan tahapan perhitungannya.
3. Gunakan persamaan yang relevan dan jelaskan arti setiap variabel.
4. Jangan membuat data lapangan yang tidak diberikan seolah-olah merupakan data sebenarnya.
5. Bedakan dengan jelas antara data, asumsi, hasil perhitungan, dan interpretasi.
6. Jika terdapat keterbatasan metode, jelaskan secara singkat.

### Output
Susun jawaban dengan format berikut:

1. Prinsip dasar metode seismik refraksi.
2. Konsep gelombang langsung dan gelombang refraksi.
3. Persamaan yang digunakan.
4. Langkah pengolahan data dari waktu tiba hingga penentuan kecepatan lapisan.
5. Contoh perhitungan kedalaman antarmuka lapisan.
6. Interpretasi kondisi bawah permukaan.
7. Kelebihan dan keterbatasan metode.
8. Kesimpulan singkat.

### Quality Criteria
Jawaban dianggap berkualitas apabila:

- Konsep seismik refraksi dijelaskan dengan benar.
- Hubungan antara kemiringan kurva waktu-tempuh dan kecepatan dijelaskan dengan tepat.
- Persamaan yang digunakan konsisten dengan konsep fisika gelombang seismik.
- Satuan setiap besaran diperhatikan.
- Tidak ada data yang dibuat-buat tanpa diberi label sebagai asumsi atau contoh.
- Interpretasi tidak melebihi informasi yang dapat didukung oleh data.
- Penjelasan dapat dipahami oleh mahasiswa S1 dan tetap menggunakan istilah geofisika yang tepat.

### Verification
Sebelum memberikan jawaban akhir, lakukan pemeriksaan berikut:

1. Periksa kembali setiap persamaan dan satuannya.
2. Pastikan kecepatan lapisan kedua lebih besar daripada lapisan pertama untuk menghasilkan refraksi kritis pada model dua lapisan yang digunakan.
3. Periksa apakah hasil kedalaman memiliki satuan meter dan masuk akal terhadap konfigurasi survei.
4. Bandingkan hasil dengan prinsip dasar metode seismik refraksi.
5. Tandai bagian yang merupakan asumsi jika data lapangan belum lengkap.
6. Jika hasil tidak konsisten secara fisik, jelaskan kemungkinan penyebabnya dan jangan memaksakan kesimpulan.

```python
import numpy as np
import matplotlib.pyplot as plt

# Contoh data ilustrasi
jarak_m = np.array([5, 10, 15, 20, 25, 30, 35, 40], dtype=float)
waktu_s = np.array([0.010, 0.020, 0.030, 0.040,
                    0.043, 0.046, 0.049, 0.052], dtype=float)

# Memeriksa data
print("Jumlah data:", len(jarak_m))
print("Jarak geofon (m):", jarak_m)
print("Waktu tiba (s):", waktu_s)

# Membuat grafik waktu tiba terhadap jarak
plt.figure(figsize=(8, 5))
plt.scatter(jarak_m, waktu_s)
plt.xlabel("Jarak dari sumber (m)")
plt.ylabel("Waktu tiba (s)")
plt.title("Grafik Waktu Tempuh Seismik Refraksi")
plt.grid(True)
plt.show()
```
