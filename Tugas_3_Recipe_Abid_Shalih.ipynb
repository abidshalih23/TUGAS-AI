# BAGIAN I --- RECIPE DARI BUKU

## 1. Recipe Audit Station CSV

### Problem

Baca satu file CSV stasiun dan identifikasi masalah pada skema kolom,
rentang waktu, nilai hilang, serta timestamp duplikat. Hasilnya berupa
ringkasan QA/QC, bukan analisis iklim atau prediksi hujan. File sumber
tidak boleh diubah.

### Minimal Theory

Audit data dilakukan sebelum agregasi atau analisis agar kesalahan
struktur, waktu, dan nilai dapat diketahui lebih awal. Timestamp ganda
dapat memengaruhi agregasi, sedangkan nilai hilang atau tipe kolom yang
tidak sesuai dapat mengubah hasil analisis.

### Input dan Constraints

-   `CSV_PATH`: path file CSV.
-   `TIME_COLUMN`: nama kolom waktu.
-   `UNITS`: kontrak satuan untuk kolom yang relevan.
-   Pertahankan file sumber; gunakan pembacaan *read-only*.
-   Jangan menghapus atau memperbaiki baris sumber secara otomatis.
-   Jangan menyimpulkan hasil ilmiah yang tidak diuji oleh audit.

### Prompt dan Procedure

> Buat draft recipe untuk audit QA/QC station CSV. Periksa skema kolom,
> rentang waktu, nilai hilang, timestamp duplikat, dan tipe data tanpa
> mengubah input mentah. Nyatakan input, teori minimum, asumsi,
> prosedur, output yang diharapkan, rencana implementasi Python,
> pemeriksaan kegagalan, serta status eksekusi. Pisahkan spesifikasi
> recipe dari klaim bahwa program telah dijalankan.

Langkah kerja: 1. Baca CSV secara *read-only*. 2. Periksa apakah kolom
wajib tersedia. 3. Validasi tipe data dan rentang timestamp. 4. Hitung
nilai hilang dan timestamp duplikat. 5. Buat laporan QA/QC dan catatan
pemeriksaan. 6. Catat error atau kolom yang tidak sesuai tanpa mengubah
sumber.

### Code / Implementation Plan

Implementasi dapat menggunakan Python dengan `pandas` untuk membaca CSV
dan menghitung ringkasan. Kode disimpan pada path relatif, misalnya
`code/recipes/io/r-io-001.py`. Path ini merupakan rancangan, bukan bukti
bahwa file sudah dibuat.

### Output

-   `results/io/station-summary.md`: laporan Markdown.
-   `results/run-records/r-io-001.json`: run record jika program
    dijalankan.

Laporan mencakup jumlah baris dan kolom, kolom yang ditemukan, rentang
waktu, jumlah nilai hilang, jumlah timestamp duplikat, dan status setiap
pemeriksaan.

### Optional Check / Evidence

Periksa bahwa file input tidak berubah, laporan tidak kosong, jumlah
baris sesuai dengan sumber, dan status eksekusi dicatat. Gunakan
`Not yet executed` jika program belum dijalankan. Label `Executed` hanya
digunakan jika bukti eksekusi tersedia.

## 2. Recipe Normalisasi GeoJSON

### Problem

Susun recipe untuk mengubah file GeoJSON menjadi GeoJSON yang
dinormalisasi, inventaris CSV, dan ringkasan JSON. Recipe harus
memeriksa geometri, properti, geometri kosong, serta urutan koordinat
tanpa menebak informasi yang tidak tersedia.

### Minimal Theory

GeoJSON menggunakan struktur JSON untuk menyimpan fitur geografis.
Pemeriksaan perlu mencakup struktur `FeatureCollection`, tipe geometri,
koordinat, properti, dan sistem referensi koordinat. Untuk koordinat
geografis GeoJSON, urutan koordinat umumnya berupa bujur lalu lintang.

### Input dan Constraints

-   `GEOJSON_PATH`: path GeoJSON sumber.
-   Path output untuk GeoJSON, CSV inventaris, dan JSON ringkasan.
-   Jangan menghilangkan fitur atau properti tanpa aturan yang jelas.
-   Jangan menebak CRS atau mengubah urutan koordinat tanpa dasar.
-   Keberhasilan membaca file tidak otomatis membuktikan topologi
    geometri valid.

### Prompt dan Procedure

> Susun recipe untuk menormalisasi `[GEOJSON_PATH]` menjadi GeoJSON,
> inventaris CSV, dan ringkasan JSON. Periksa struktur, tipe geometri,
> properti, geometri kosong atau tidak valid, urutan koordinat, dan
> informasi CRS. Nyatakan ketidakpastian secara terbuka dan jangan
> mengklaim output telah dihasilkan sebelum eksekusi dibuktikan.

Langkah kerja: 1. Baca dan validasi sintaks JSON. 2. Periksa struktur
objek dan daftar fitur. 3. Periksa geometri, koordinat, serta properti.
4. Identifikasi CRS atau urutan koordinat yang belum jelas. 5.
Normalisasi hanya berdasarkan aturan yang terdefinisi. 6. Ekspor
inventaris fitur dan ringkasan hasil pemeriksaan. 7. Catat fitur yang
gagal diproses.

### Code / Implementation Plan

Implementasi dapat menggunakan Python dengan pustaka yang mendukung
GeoJSON, misalnya `json` untuk pemeriksaan struktur dan pustaka
geospasial yang tersedia di lingkungan kerja. Dependency harus dicatat
dan tidak boleh diasumsikan sudah terpasang.

### Output

-   GeoJSON hasil normalisasi.
-   CSV inventaris fitur.
-   JSON ringkasan QA/QC.
-   Run record serta peringatan jika tersedia.

### Optional Check / Evidence

Bandingkan jumlah fitur sumber dan hasil, periksa properti penting,
validasi struktur output, dan catat fitur yang tidak berhasil diproses.
Jika program belum dijalankan, sebutkan output sebagai hasil yang

## 3. Recipe Model Difusi Panas Satu Dimensi

### Problem

Buat recipe untuk mensimulasikan perubahan suhu terhadap posisi dan
waktu berdasarkan profil awal, difusivitas termal, kondisi batas, jarak
grid, dan langkah waktu.

### Minimal Theory

Persamaan difusi panas satu dimensi adalah:

\[
\frac{\partial T}{\partial t}=\alpha\frac{\partial^2T}{\partial x^2}
\]

dengan (T) adalah suhu, (t) waktu, (x) posisi, dan (\alpha)
difusivitas termal.

Untuk skema beda hingga eksplisit standar, parameter kestabilan dapat
ditulis:

\[ r=\frac{\alpha\Delta t}{(\Delta x)^2} \]

Pada kasus satu dimensi dengan skema eksplisit standar, syarat
kestabilan yang umum digunakan adalah (r \le 1/2). Syarat ini
perlu disesuaikan dengan skema numerik yang benar-benar digunakan.

### Input dan Constraints

-   `INITIAL_PROFILE`: profil suhu awal.
-   `ALPHA`: difusivitas termal dan satuannya.
-   `BOUNDARIES`: kondisi batas.
-   `DZ`: jarak grid.
-   `DT`: langkah waktu.
-   Pastikan konsistensi satuan.
-   Jangan mengklaim kalibrasi atau konvergensi tanpa pengujian.

### Prompt dan Procedure

> Buat draft recipe model difusi panas satu dimensi menggunakan profil
> awal `[INITIAL_PROFILE]`, difusivitas `[ALPHA]`, kondisi batas
> `[BOUNDARIES]`, jarak grid `[DZ]`, dan langkah waktu `[DT]`. Sertakan
> persamaan pengatur, konvensi satuan dan tanda, pemeriksaan kestabilan,
> pembanding analitik atau kasus dasar, profil output, grafik, run
> record, serta peringatan jika parameter atau dependency belum
> tersedia. Jangan menyatakan eksekusi tanpa bukti.

Langkah kerja: 1. Validasi parameter dan satuan. 2. Bentuk grid ruang
dan waktu. 3. Terapkan kondisi awal serta kondisi batas. 4. Periksa
kestabilan skema. 5. Jalankan iterasi jika implementasi tersedia. 6.
Bandingkan hasil dengan kasus dasar atau solusi analitik jika
memungkinkan. 7. Simpan profil suhu, grafik, dan catatan eksekusi.

### Code / Implementation Plan

Implementasi dapat menggunakan Python, `numpy` untuk komputasi numerik,
dan `matplotlib` untuk visualisasi. Pilihan metode numerik, parameter,
serta dependency harus dicatat.

### Output

-   Profil suhu terhadap posisi dan waktu.
-   Grafik hasil simulasi.
-   Ringkasan parameter.
-   Run record berisi status dan peringatan.

### Optional Check / Evidence

Periksa kestabilan numerik, kondisi batas, konsistensi satuan, dan
kesesuaian dengan kasus dasar. Jangan mengklaim konvergensi, kalibrasi,
atau keberhasilan eksekusi tanpa bukti yang sesuai.

# BAGIAN II --- RECIPE BARU BUATAN SAYA

## 4. Recipe QA/QC dan Interpretasi Data Seismik Refraksi

### Problem

Menyusun workflow yang dapat membaca data waktu tempuh gelombang pertama (*first arrival*) dari survei seismik refraksi, memeriksa kualitas data, memvisualisasikan kurva waktu tempuh terhadap jarak, dan membuat estimasi awal kecepatan lapisan bawah permukaan. Recipe ini berfokus pada pemeriksaan data dan interpretasi awal model dua lapisan, bukan klaim hasil survei tertentu.

### Minimal Theory

Pada survei seismik refraksi, waktu tempuh gelombang pertama direkam pada sejumlah geofon dengan jarak tertentu dari sumber. Untuk lapisan homogen, hubungan waktu tempuh gelombang langsung secara sederhana dapat ditulis:

\[
t = \frac{x}{V_1}
\]

dengan \(t\) adalah waktu tempuh, \(x\) jarak sumber-geofon, dan \(V_1\) kecepatan lapisan atas.

Jika kecepatan lapisan bawah lebih besar daripada lapisan atas (\(V_2 > V_1\)), gelombang kepala yang dibiaskan secara kritis dapat muncul pada jarak tertentu. Pada grafik waktu tempuh terhadap jarak, kemiringan segmen garis berkaitan dengan kebalikan kecepatan semu:

\[
V_{\mathrm{app}} = \frac{1}{m}
\]

dengan \(m\) adalah kemiringan garis waktu tempuh terhadap jarak. Interpretasi dua lapisan menggunakan pendekatan ini hanya berlaku jika asumsi model sesuai dengan kondisi geologi dan geometri survei.

### Input dan Constraints

Input utama:
- `DATA_PATH`: lokasi file CSV waktu tempuh.
- `SHOT_ID`: identitas titik tembak.
- `OFFSET_M`: jarak sumber ke setiap geofon dalam meter.
- `FIRST_ARRIVAL_MS`: waktu tiba pertama dalam milidetik.
- `GEOHONE_ID`: identitas geofon.

Input tambahan jika tersedia:
- `SHOT_DIRECTION`: arah atau posisi sumber tembak.
- `ELEVATION_M`: elevasi sumber dan geofon.
- `PICK_UNCERTAINTY_MS`: ketidakpastian picking waktu tiba.

Batasan:
1. File mentah dibaca tanpa mengubah sumber.
2. Jarak dan waktu harus numerik serta menggunakan satuan yang konsisten.
3. Waktu tiba negatif, nilai kosong, dan identitas geofon duplikat harus ditandai untuk diperiksa.
4. Data tidak boleh dihapus otomatis hanya karena tampak menyimpang; setiap pengecualian harus dicatat dan memiliki alasan.
5. Kecepatan lapisan tidak boleh ditafsirkan dari satu titik saja; diperlukan segmen data yang cukup dan pola waktu-tempuh yang dapat dipertanggungjawabkan.
6. Estimasi dua lapisan tidak membuktikan kondisi geologi unik dan harus disertai asumsi serta keterbatasan.

### Prompt dan Procedure

**Prompt yang dapat digunakan kembali:**

> Bertindaklah sebagai asisten pengolahan data seismik refraksi. Baca `[DATA_PATH]` secara read-only dan periksa kolom `GEOHONE_ID`, `OFFSET_M`, serta `FIRST_ARRIVAL_MS`. Audit nilai kosong, tipe data, nilai waktu yang tidak masuk akal, duplikasi geofon, dan urutan offset. Buat tabel QA/QC dan grafik waktu tiba pertama terhadap offset. Jika jumlah dan pola data memadai, estimasikan kemiringan segmen garis untuk memperoleh kecepatan semu awal. Nyatakan asumsi pemilihan segmen, satuan, ketidakpastian, dan status eksekusi. Jangan mengklaim kedalaman atau model lapisan yang tervalidasi tanpa analisis tambahan.

**Procedure:**
1. Baca CSV secara read-only dan catat ukuran serta nama kolom.
2. Pastikan kolom wajib tersedia dan ubah data numerik hanya pada salinan kerja.
3. Periksa nilai kosong, nilai nonnumerik, waktu negatif, duplikasi ID geofon, dan offset berulang.
4. Urutkan salinan kerja berdasarkan offset dan simpan daftar anomali tanpa mengubah sumber.
5. Plot first-arrival time terhadap offset; tampilkan satuan pada kedua sumbu.
6. Identifikasi segmen garis yang secara fisik masuk akal, dengan pemilihan segmen dicatat secara eksplisit.
7. Estimasikan kemiringan tiap segmen melalui regresi linear dan hitung kecepatan semu sebagai kebalikan kemiringan setelah satuan diseragamkan.
8. Jika data dan asumsi mendukung, bandingkan segmen gelombang langsung dan gelombang refraksi sebagai interpretasi awal dua lapisan.
9. Laporkan residual atau ukuran kecocokan regresi, keterbatasan, dan status eksekusi. Jangan membuat klaim kedalaman lapisan jika parameter dan asumsi yang diperlukan belum tersedia.

### Code / Implementation Plan

Implementasi dapat menggunakan Python:
- `pandas`: membaca CSV dan melakukan audit tabel.
- `numpy`: operasi numerik dan pengurutan data.
- `matplotlib`: membuat grafik waktu tempuh terhadap offset.
- `scipy.stats` atau `numpy.polyfit`: regresi linear untuk segmen yang telah dipilih.
- `json` dan `pathlib`: menyimpan konfigurasi serta run record.

Konversi satuan harus dilakukan secara eksplisit. Jika waktu disimpan dalam milidetik, ubah ke detik sebelum menghitung kemiringan dalam s/m agar kecepatan yang diperoleh memiliki satuan m/s.

Rancangan struktur direktori:

```text
seismic-refraction-qc/
├── data/
│   └── first-arrivals.csv
├── code/
│   └── refraction_qc.py
└── results/
    ├── refraction-qc-report.md
    ├── travel-time-plot.png
    ├── segment-fit.csv
    └── run-record.json
```

Struktur tersebut merupakan rancangan dan bukan bukti bahwa file telah dibuat atau program telah dijalankan.

### Output

1. `refraction-qc-report.md`: ringkasan skema data, jumlah baris, data hilang, duplikasi, anomali, dan catatan asumsi.
2. `travel-time-plot.png`: grafik waktu tiba pertama terhadap offset, dengan data terukur dan garis regresi yang dibedakan.
3. `segment-fit.csv`: daftar segmen yang dianalisis, jumlah titik, kemiringan, kecepatan semu, dan ukuran kecocokan regresi.
4. `run-record.json`: waktu eksekusi, konfigurasi, versi dependency, status, serta pesan kesalahan jika ada.

### Optional Check / Evidence

Checklist pemeriksaan:
- [ ] File sumber tetap tidak berubah.
- [ ] Kolom wajib dan satuan telah diperiksa.
- [ ] Data kosong, nonnumerik, offset duplikat, dan waktu negatif telah ditandai.
- [ ] Grafik menggunakan offset dalam meter dan waktu dalam satuan yang jelas.
- [ ] Konversi milidetik ke detik dilakukan sebelum menghitung kecepatan.
- [ ] Pemilihan segmen regresi dijelaskan dan dapat diulang.
- [ ] Kecepatan semu tidak langsung dianggap sebagai kecepatan lapisan sebenarnya.
- [ ] Status eksekusi sesuai bukti yang tersedia.

Status eksekusi:
- `Draft`: spesifikasi recipe sedang dirancang.
- `Not yet executed`: program belum dijalankan.
- `Partial`: audit data dilakukan, tetapi estimasi atau pemeriksaan lain belum selesai.
- `Executed`: program dijalankan dan output yang dinyatakan benar-benar tersedia.

### Keterbatasan dan Risiko

Interpretasi seismik refraksi dipengaruhi oleh ketelitian picking waktu tiba, geometri sumber dan geofon, topografi, variasi lateral, serta asumsi struktur lapisan. Kecepatan semu dari kemiringan grafik tidak selalu sama dengan kecepatan sebenarnya. Model sederhana dua lapisan dapat gagal pada kondisi lapisan berkecepatan rendah di bawah lapisan berkecepatan tinggi, batas lapisan miring, atau struktur yang heterogen. Karena itu, hasil recipe merupakan pemeriksaan dan estimasi awal, bukan pengganti evaluasi geofisika menyeluruh.

### Ringkasan Recipe Baru

Recipe ini menyediakan alur yang dapat diulang untuk mengaudit data first-arrival seismik refraksi, menampilkan kurva waktu tempuh terhadap offset, serta mengestimasi kecepatan semu melalui regresi linear pada segmen yang dipilih. Pemisahan data mentah, hasil perhitungan, asumsi, dan keterbatasan membantu menjaga transparansi serta mencegah interpretasi lapisan yang tidak didukung data.
